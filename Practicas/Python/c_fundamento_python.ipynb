### Logic and comparition operators on python

```python
# Comparition operators

x = 5
y = 6

print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

```python
# Logic operators
# allows me to perform multiple logic validations

x = 5
y = 2
z = 10

# And
"""
Logic operator AND work comparing two or more logic conditions
if any of these is false all logic result should be FALSE
if all of them are TRUE the result should be TRUE 
"""
print(x > y and z < y)

# Or
"""
Logic operator OR works comparing two or more logic conditions
if any of these is TRUE all the result should be TRUE as well
only when all logic tests are FALSE the result will be FALSE
"""
print(x > y or z < y)

# Not
"""
This logic operator allows me to change the boolean state
"""
print("Negacion:", not(True))
```

### Conditionals on python, if, else, and or and pass

```python
# if conditional
"""
This sentence allows me to execute a specific block of code depending of a logic evaliation
"""

x = 5
y = 3
z = 3

if x > y and x > z:
    print("X es mayor")
elif y > x or y > z:
    print("Y es mayor")
else:
    print("son iguales")

```

```python
# Use of "pass" sentence
"""
This sentence allows me to avoid an error defining conditionals as if while I am programing or developong
the logic in this sentence os specific block of code
"""

if 5 > 3:
    print("hello world")
else:
    pass
```

### MATCH sentence on python 3.10 

THis sentence is similar to SWITCH in other programing languages

```python
# MATCH sentence

x = 2

match x:
    case 1:
        print("Lunes")
    case 2:
        print("Martes")
    case _:
        print("No coincide con las opciones") # case _ catch all the cases that doesn't fit with other cases
```

### While Loops on python: conditions, break and continue

While repeating sentences while aa specific condition been true

```python
# While loop

x = 1

while x < 10:
    x += 1
    if x == 5:
        print("X:", 5)
        continue
    elif x == 9:
        break
    print(x)
```

### For loop on python

This sentence si usefull to loop through a string or a list 

```python
# for loop

word = "python"

for i in word:
    print(i)
```

```python
# for loop

list = ["Apple", "Orange", "Mango"]

for fruit in list:
    if fruit == "Orange":
        continue
    print(fruit)
else:
    print("Program ends")
```

```python
# Use of for and range(start, stop, step)
"""
start by default is cero
step by default is one
"""

for i in range(2,10,2):
    print(i)
```

```python
fruits = ["Apple", "Orange", "Mango"]
adjetives = ["Jummy", "fresh"]

for fruit in fruits:
    for adj in adjetives:
        print(f'{fruit} is {adj}')
```

## Data structure fundamentals

### Esential method from lists on Python

These are collections of different elements that can be stored in only one varibale a list
Lists are  sorted, unmutables and can store heterogeneal data

```python
# list

fruits = ["apple", "orange", "tangerine"]
print(fruits)
```

```python
# List are sorted, it means that I can access or change a specific value by it index

print(fruits[1])
fruits[1] = "Banana"
print(fruits[1])
```

```python
# List can has duplicated elements

fruits.append("orange")
print(fruits)
```

```python
# How can I know how many elements has a list

print(len(fruits))
```

```python
# I can take a range o indexes to see several elements

print(fruits[0:2]) # Index 2 is not included
```

```python
# use of in sentece to check if a specific element exist inside the list

print("orange" in fruits)

if "orange" in fruits:
    print("orange exist in fruits list")
```

#### 1. List methods

```python
# .append() method, allows me to add at the end o a list a specific new element

fruits.append("True")
print(fruits)
```

```python
# insert(index, element) this method allows me to insert a specific element in alist in a specific index
# The new element doen't remove the element in index

fruits.insert(0, "Hugo")
print(fruits)
```

```python
#.remove(element) this method allows me to remove a specific value only it first coincidence in the list

fruits.insert(3,"True")
print(fruits)
fruits.remove("True")
print(fruits)
```

```python
# .pop(index) method, this method allows me to remove a specific value with it index from a list

print(fruits)
fruits.pop(2)
print(fruits)
```

```python
# Sorting list with .sort() and .reverse() merthod
# these methods allows me to sort the list

fruits.sort()
print(fruits)

fruits.reverse()
print(fruits)

fruits.sort()
print(fruits)
```

```python
# concatenating list

l1 = [1,2,'True']
l2 = [3,False,5]

l3 = l1 + l2
print(l3)

# use of extend(list) useful to add a list to another, this method will change a existing one

l1.extend(l2)
print(l1)
```
