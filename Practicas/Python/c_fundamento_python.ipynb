## Data types on Python

#### Text type
Also called string, must be between " " , ' ' , """ """ or ''' ''' to be assign

#### Number type

1. integer int
2. floating float
3. imaginary

#### List type
Sorted and mutable collection, sorted because all it's items have an index

#### Tuple
Sorted and inmutable collection, sorted because all it's items have an index

#### Dictionary
Sorted collection of key value content and mutable, I can access the values throuht a index or the key

#### Sets
Sorted and mutable collection, sets doesn't allow repeated values

#### Booleans
This data type can has only two values TRUE or FALSE

```python
# String variables assignment
var1 = "value_1"
print(var1)
var2 = 'value_2'
print(var2)
var3 = '''value_3'''
print(var3)
var4 = """value_4"""
print(var4)
```

```python
# Number type assignment
int_var = 23
print(int_var)
float_var = 2.5
print(float_var)
imagin_var = 3 + 2j
print(imagin_var)
```

```python
# List type
'''
alist must be defined between [], it's a sorted and mutable data type, lists has an index
I can replace, remove or add any kind of data type, lists allow duplicated data
'''
list_1 = [1,2,3,"one"]
print(list_1)
# A list can be access by the value index begining from cero
print(list_1[2]) #-> Shows the 3rth element
# I can replace a specific elemnet of a list using it's index
list_1[2] = "Two"
print(list_1)
```

```python
# Tuple type
'''
A tuple is a sorted and inmutable data type, must be defined between (), it's elements have index
I can acces to it's elements by index but not change it I cannot replace, remove or add elements
'''
tuple_1 = (1,2,"Hello")
print(tuple_1)
# I can access tuple elements by index
print(tuple_1[2])
```

### Arithmetic operators on Python

```python
a = 12
b = 5

# addition
print("add:", a+b)

# sustraction
print("sustract:", a-b)

# multiplication
print("multipl:", a*b)

# division
print("division:", a/b)

# modulus operator
print("modulo:", a%b)

# square
print("square:", b**2)

# floor division 
print("floor:", 7 // 2) # result 3.5
print("floor:", -7 // 2) # result 3.5
# La operacion floor division realiza la division pero redondea hacia abajo
# al entero mas cercano por eso tengo el primer resultado 3 y -4
```

### Asignation Operators and WALRUS on Python

```python
# Assignation operators

x = 5
print(x)

x = x+1
print(x)
# Also I can Write like 
x += 1
print(x)

x -= 1
print(x)

x *= 2
print(x)

x /= 4
print(x)

x %= 2
print(x)

y = 20

y //= 3
print(y)

y = 10
y **= 3
print(y)

# WALRUS operator
"""
This spetial operator allows me to declarate a variable inside a method
it simbol is := and an example
"""

print(var := 10)

var += 1
print(var) # before var doesn't exists
```

### Logic and comparition operators on python

```python
# Comparition operators

x = 5
y = 6

print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

```python
# Logic operators
# allows me to perform multiple logic validations

x = 5
y = 2
z = 10

# And
"""
Logic operator AND work comparing two or more logic conditions
if any of these is false all logic result should be FALSE
if all of them are TRUE the result should be TRUE 
"""
print(x > y and z < y)

# Or
"""
Logic operator OR works comparing two or more logic conditions
if any of these is TRUE all the result should be TRUE as well
only when all logic tests are FALSE the result will be FALSE
"""
print(x > y or z < y)

# Not
"""
This logic operator allows me to change the boolean state
"""
print("Negacion:", not(True))
```
