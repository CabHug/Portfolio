### Series

Es la unidad basica de datos de pandas, pude ser creada con una lista, tupla, diccionarios a rreglos de numpy

```python
# Una serie se crea con la funcion (pd.Series)
ciudades = ['Bogotá', 'Cali', 'Medellín','Barranquilla']
serie_ciudades = pd.Series(ciudades)
serie_ciudades
```

```python
# Validamos el timpo de dato de serie_inicial
type(serie_ciudades)
```

```python
# en una serie puedo definir tanto los indices como los valores
habitantes = (11344000,2228000,2569000,1206000)

#En este punto vamos a crear una serie con sus indices
serie_habitantes = pd.Series(
    data=habitantes,
    index= ciudades
)
serie_habitantes
```

### Dataframe
Un data frame en la estructura de datos mas amplia y conocida de pandas, esta representa una tabla en donde cada columna representa una variable del estudio y cada fila o registro una medicion
El dataframe se compone de series y posee caracteristicas similares a la de una serie

```python
datos = {
    'nombre_ciudad': ['Bogotá', 'Cali', 'Medellín','Barranquilla'],
    'poblacion' : (11344000,2228000,2569000,1206000),
    'altura' : np.array([2640,1018,1495,18])
}

# En este ejemplo el objeto de estudio son las ciudades
# Y cada columna es una variable con la que podemos describir a cada ciudad
ciudades = pd.DataFrame(datos)
ciudades
```

#### Analisis de variables categoricas o cualitativas


```python
# el metodo .unique() me muestra las valores unicos que conforman las mediciones de la variables
housing['furnishingstatus'].unique()
```

```python
# Con el metodo nunique() me permite regresar el numero de elementos unicos en la lista
housing['furnishingstatus'].nunique()
```

```python
# el metodo value_counts() me permite tener un conteo de cada una de los valores en la variable
housing['furnishingstatus'].value_counts()
```

```python
# De esta forma puedo calcular el porcentaje de cada medida
numero_filas = housing.shape[0]

np.round(housing['furnishingstatus'].value_counts() / numero_filas*100 , 2)
```
