### Resumen estadistico del set de datos
En este resumen estadistico general donde se muestra diferentes medidas estadisticas de las cuales resaltan:
- count: muestra un reconteo de el numero de registros que se encuentran en cada columna o serie
- std: desviacion estandar me muestra que tan desviados estan los datos respecto de su media
- min y max: que permite ver o calcular el rango en el que se encuentran los datos cuantitativos en este dataframe

#### Revision de la variable price:
- La variable price tiene un ranto de ($0 a $500)
- Dado el Q2 = 100 un valor min = 0 y max = 500, se puede concluir que la distribusion de los datos de proce tiene una distribusion positiva o un sesgo positivo
- Para encontrar si tiene valores atipicos calculo el rango intercuartilico 
    IQR = 175 - 65 = 110
    rango inferior = Q1 - (1.5 * IQR)
    rango superior = Q3 + (1.5 * IQR)
    rango superior = 175 + 165 = 340
    Con esto podemos concluir que para lo valores mayores a $285 se pueden considerar como outliers
- Los valores son logicos dentro del negocio, es necesario indagar un poco mas en los valores atipicos para encontrar si corresponden a espacion amplios o ubicaciones de alto costo

#### Analisis de la variable neighbourhood

```python
neighb = AB_df['neighbourhood'].value_counts().head(10)
neighb_df = neighb.reset_index()
neighb_df.columns= ['neighbourhood','count']
print(pd.DataFrame(neighb))


sns.barplot(
    data=neighb_df,
    x='neighbourhood',
    y='count'
)

plt.xticks(rotation=45)
plt.title('Top 10 barrios en Airbnb NYC')
plt.show()
```
