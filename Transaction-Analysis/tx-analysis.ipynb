```python
# ¿Cuántas transacciones hay en total y cuántas por status (APPROVED/DECLINED)?

query ="""
SELECT
    tx.status as status,
    COUNT(tx.tx_date)
FROM transactions as tx
GROUP BY
    status

UNION ALL

SELECT
    'TOTAL' as status,
    COUNT(tx.tx_date) as total
FROM transactions as tx
"""
rslt = spark.sql(query).show()
```

```python
# ¿Cuál es el monto total y el promedio transaccionado por cada merchant?

query = """
SELECT
    mr.merchant_name as merchant,
    SUM(tx.amount) as amt,
    ROUND(MEAN(tx.amount), 2) as mean
FROM transactions as tx
LEFT JOIN merchants as mr
    on mr.merchant_id = tx.merchant_id
GROUP BY
    merchant
"""
rslt = spark.sql(query).show()
```
